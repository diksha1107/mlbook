```python
from datetime import datetime, timedelta

today = datetime.today()

# Force "this week's Friday" within same Mon–Sun window
this_friday = today - timedelta(days=(today.weekday() - 4))

last_friday = this_friday - timedelta(days=7)

print("This week's Friday:", this_friday.date())
print("Last week's Friday:", last_friday.date())
```
